### Exercice 1 : Le problème linéaire (1 neurone suffit)

**Instructions :**
1. Ouvrez [playground.tensorflow.org](https://playground.tensorflow.org)
2. Sélectionnez le dataset **"Circle"** (en haut à gauche)
3. Ne gardez qu'**UN SEUL neurone** dans la couche cachée
4. Cliquez sur **Play** (▶️)

**Question :** Que se passe-t-il ? Le réseau arrive-t-il à séparer les deux couleurs ?

<details>
<summary>🔑 Réponse Attendue</summary>

**Non !** Un seul neurone ne peut tracer qu'une **ligne droite**. 

Le dataset "Circle" nécessite une frontière **courbe** (un cercle). Une ligne droite ne peut pas séparer un cercle en deux.

Vous devriez voir :
- La loss qui reste élevée
- Une ligne droite qui "essaie" de séparer, sans succès

**Conclusion :** Pour des patterns non-linéaires, un seul neurone ne suffit pas.

</details>

### Exercice 2 : Ajouter des neurones (largeur)

**Instructions :**
1. Restez sur le dataset **"Circle"**
2. Augmentez à **4 neurones** dans la couche cachée
3. Relancez l'entraînement

**Question :** Le réseau arrive-t-il maintenant à séparer les couleurs ? Pourquoi ?

<details>
<summary>🔑 Réponse Attendue</summary>

**Oui !** Avec 4 neurones, le réseau peut apprendre à tracer une frontière circulaire.

Explication :

Observez les visualisations de chaque neurone — chacun détecte une "partie" du cercle.

### Exercice 3 : Le problème de la spirale (besoin de profondeur)

**Instructions :**
1. Sélectionnez le dataset **"Spiral"** (le plus difficile !)
2. Commencez avec 1 couche de 4 neurones
3. Observez : le réseau converge-t-il ?

**Question :** Que faut-il faire pour résoudre la spirale ?

<details>
<summary>🔑 Réponse Attendue</summary>

Avec 1 couche, même avec beaucoup de neurones, la spirale est TRÈS difficile.

**Solution :** Ajouter des COUCHES (profondeur) !

Essayez :
- 2 couches de 6 neurones chacune
- Activation : **ReLU** (pas tanh ou sigmoid)
- Learning rate : 0.03 ou 0.1

Vous devriez voir le réseau "dérouler" la spirale progressivement.

**Conclusion :** La spirale nécessite des abstractions HIÉRARCHIQUES — c'est pourquoi la profondeur est nécessaire.

</details>

### Exercice 4 : Comparer les activations

**Instructions :**
1. Restez sur la spirale avec 2 couches de 6 neurones
2. Testez avec **Sigmoid**, puis **Tanh**, puis **ReLU**
3. Observez la vitesse de convergence et la loss finale

**Question :** Quelle activation fonctionne le mieux ? Pourquoi ?

<details>
<summary>🔑 Réponse Attendue</summary>

**ReLU** converge généralement le plus vite et atteint la loss la plus basse.

Raisons :
- **Sigmoid/Tanh** : Vanishing gradient — le réseau apprend lentement car les gradients s'évaporent
- **ReLU** : Gradient = 1 pour les valeurs positives — l'information passe bien

Vous pouvez observer la différence en regardant combien d'epochs sont nécessaires pour atteindre une loss basse.

</details>

### Exercice 5 : Observer l'overfitting

**Instructions :**

2. Créez un réseau TRÈS large : 3 couches de 8 neurones

**Question :** Que remarquez-vous sur la frontière ? Est-ce bon signe ?

<details>
<summary>🔑 Réponse Attendue</summary>

La frontière devient **très tortueuse** et essaie de capturer chaque point individuellement.

C'est de l'**OVERFITTING** :
- Le réseau mémorise les points de bruit

- Sur de nouvelles données, il se tromperait

**Solutions :**
- Réduire la taille du réseau
- Ajouter de la régularisation (Dropout)
- Avoir plus de données

## 6.2 Keras — Votre premier réseau de neurones en Python

Maintenant que vous avez visualisé les concepts, écrivons du vrai code !

Nous allons classifier des chiffres manuscrits (MNIST) — le "Hello World" du Deep Learning.

### Installation et imports

````python
# Étape 1 : Importer les bibliothèques
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt

# Vérifier la version de TensorFlow
print(f"TensorFlow version: {tf.__version__}")
````

### Charger le dataset MNIST

MNIST contient 70,000 images de chiffres manuscrits (0-9), chacune de 28×28 pixels.

````python
# Étape 2 : Charger les données
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

# Afficher les dimensions
print(f"X_train shape: {X_train.shape}")  # (60000, 28, 28)
print(f"y_train shape: {y_train.shape}")  # (60000,)
print(f"X_test shape: {X_test.shape}")    # (10000, 28, 28)
print(f"y_test shape: {y_test.shape}")    # (10000,)
````

**Question :** Que représentent ces dimensions ?

*(Réponse attendue : 60,000 images d'entraînement, chacune de 28×28 pixels. 10,000 images de test.)*

````python
# Étape 3 : Visualiser quelques exemples
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train[i], cmap='gray')
    ax.set_title(f"Label: {y_train[i]}")
    ax.axis('off')
plt.suptitle("Exemples du dataset MNIST", fontsize=14)
plt.tight_layout()
plt.show()
````

### Préparer les données

Avant de donner les données au réseau, nous devons :
1. **Aplatir** les images (28×28 → 784 pixels)
2. **Normaliser** les valeurs (0-255 → 0-1)

````python
# Étape 4 : Prétraitement des données

# Aplatir : 28x28 → 784
X_train_flat = X_train.reshape(60000, 784)
X_test_flat = X_test.reshape(10000, 784)

# Normaliser : 0-255 → 0-1
X_train_norm = X_train_flat / 255.0
X_test_norm = X_test_flat / 255.0

print(f"Après prétraitement: {X_train_norm.shape}")
print(f"Min: {X_train_norm.min()}, Max: {X_train_norm.max()}")
````

**Question :** Pourquoi normaliser les données entre 0 et 1 ?

*(Réponse attendue : Les valeurs trop grandes peuvent causer des problèmes numériques. Normaliser aide le réseau à converger plus vite et de manière plus stable.)*

### Construire le modèle MLP

Voici le moment clé ! Nous allons créer un MLP avec :
- 784 entrées (pixels)
- 128 neurones (couche cachée 1) + ReLU
- 64 neurones (couche cachée 2) + ReLU
- 10 sorties (chiffres 0-9) + Softmax

````python
# Étape 5 : Créer le modèle

model = keras.Sequential([
    # Couche d'entrée : 784 features (pixels aplatis)
    layers.Input(shape=(784,)),
    
    # Couche cachée 1 : 128 neurones, activation ReLU
    layers.Dense(128, activation='relu'),
    
    # Couche cachée 2 : 64 neurones, activation ReLU
    layers.Dense(64, activation='relu'),
    
    # Couche de sortie : 10 neurones (classes 0-9), activation Softmax
    layers.Dense(10, activation='softmax')
])

# Afficher l'architecture
model.summary()
````

### Comment lire le summary ?

```

```

### Compiler le modèle

Avant l'entraînement, nous devons spécifier :
- **Optimizer** : Comment ajuster les poids (Adam = Gradient Descent amélioré)
- **Loss** : Comment mesurer l'erreur (Cross-Entropy pour classification)
- **Metrics** : Ce qu'on veut surveiller (accuracy)

````python
# Étape 6 : Compiler le modèle

model.compile(
    optimizer='adam',                        # Optimizer adaptatif (meilleur que SGD simple)
    loss='sparse_categorical_crossentropy',  # Loss pour classification multi-classe
    metrics=['accuracy']                     # Métrique à afficher
)

print("Modèle compilé avec succès !")
````

### Entraîner le modèle

C'est ici que la magie opère ! Le modèle va :
1. Faire des Forward Pass
2. Calculer la Loss
3. Faire la Backpropagation
4. Mettre à jour les poids

...des milliers de fois !

````python
# Étape 7 : Entraîner le modèle

history = model.fit(
    X_train_norm,      # Données d'entraînement
    y_train,           # Labels
    epochs=10,         # Nombre de passages sur les données
    batch_size=32,     # Taille des mini-batches
    validation_split=0.1,  # 10% pour validation
    verbose=1          # Afficher la progression
)
````

**Question :** Observez la sortie. Que signifient `loss`, `accuracy`, `val_loss`, `val_accuracy` ?

*(Réponse attendue :*
- *loss = erreur sur le train set*
- *accuracy = % de bonnes prédictions sur train*
- *val_loss = erreur sur le validation set (10% mis de côté)*
- *val_accuracy = % de bonnes prédictions sur validation*

*Si val_loss remonte alors que loss descend → overfitting !)*

### Visualiser l'entraînement

````python
# Étape 8 : Visualiser les courbes d'apprentissage

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Courbe de Loss
axes[0].plot(history.history['loss'], label='Train Loss')
axes[0].plot(history.history['val_loss'], label='Validation Loss')
axes[0].set_title('Loss pendant l\'entraînement')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(True)

# Courbe d'Accuracy
axes[1].plot(history.history['accuracy'], label='Train Accuracy')
axes[1].plot(history.history['val_accuracy'], label='Validation Accuracy')
axes[1].set_title('Accuracy pendant l\'entraînement')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
````

### Évaluer sur le test set

````python
# Étape 9 : Évaluer sur les données de test

test_loss, test_accuracy = model.evaluate(X_test_norm, y_test, verbose=0)

print(f"\n" + "="*50)
print(f"RÉSULTATS SUR LE TEST SET")
print(f"="*50)
print(f"Loss: {test_loss:.4f}")
print(f"Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"="*50)
````

**Félicitations !** Vous avez entraîné un réseau de neurones qui reconnaît les chiffres manuscrits avec ~97% de précision ! 

### Faire des prédictions

````python
# Étape 10 : Faire des prédictions sur quelques exemples

# Prédire sur les 5 premières images du test set
predictions = model.predict(X_test_norm[:5])

# Afficher les résultats
fig, axes = plt.subplots(1, 5, figsize=(15, 3))
for i, ax in enumerate(axes):
    ax.imshow(X_test[i], cmap='gray')
    predicted_class = np.argmax(predictions[i])
    confidence = predictions[i][predicted_class] * 100
    true_label = y_test[i]
    
    color = 'green' if predicted_class == true_label else 'red'
    ax.set_title(f"Prédit: {predicted_class} ({confidence:.1f}%)\nVrai: {true_label}", color=color)
    ax.axis('off')

plt.suptitle("Prédictions du modèle", fontsize=14)
plt.tight_layout()
plt.show()
````

## 6.3 Synthèse & Vocabulaire

### Récapitulatif du vocabulaire technique

Après ce chapitre, vous devez maîtriser ces termes :

| Terme | Définition |
|-------|------------|
| **Neurone** | Unité de calcul : somme pondérée + activation |
| **Poids (weights)** | Paramètres appris qui pondèrent les entrées |
| **Biais (bias)** | Paramètre appris qui décale l'activation |
| **Activation** | Fonction non-linéaire (ReLU, Sigmoid, Softmax) |
| **MLP** | Multi-Layer Perceptron : réseau fully connected |
| **Dense** | Couche fully connected (synonyme) |
| **Forward Pass** | Calcul de la prédiction |
| **Loss** | Mesure de l'erreur |
| **Backpropagation** | Calcul des gradients |
| **Gradient Descent** | Optimisation par petits pas |
| **Learning Rate** | Taille des pas |
| **Epoch** | Un passage complet sur les données |
| **Batch** | Sous-ensemble pour une mise à jour |
| **Dropout** | Régularisation par extinction de neurones |
| **Batch Normalization** | Normalisation pour stabilité |
