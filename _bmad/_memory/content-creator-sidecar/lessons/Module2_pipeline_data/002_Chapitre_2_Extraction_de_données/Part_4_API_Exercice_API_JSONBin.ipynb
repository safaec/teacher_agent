## 📤 Étape 2 : Créer un "Bin" avec POST

Un **Bin** est un conteneur JSON stocké sur JSONBin.io.

### Ce que fait cette requête :
- **Méthode** : POST (créer une nouvelle ressource)
- **URL** : `https://api.jsonbin.io/v3/b`
- **Headers** :
  - `Content-Type: application/json` → On envoie du JSON
  - `X-Master-Key: <ta_clé>` → Authentification
  - `X-Bin-Private: false` → Rendre le bin public (visible dans le navigateur !)
- **Body** : Les données JSON à stocker

```python
# ============================================
# 📤 CRÉER UN NOUVEAU BIN (POST)
# ============================================

# Les données que tu veux stocker
mon_profil = {
    "nom": "Ton Nom",           # ← Mets ton vrai nom !
    "prenom": "Ton Prénom",     # ← Mets ton vrai prénom !
    "formation": "Python & APIs",
    "niveau": "Débutant",
    "competences": ["Python", "HTTP", "REST", "JSON"],
    "date_creation": datetime.now().isoformat(),
    "message": "Mon premier bin créé via l'API ! 🎉"
}

# Headers spéciaux pour cette requête
headers_creation = {
    **HEADERS,  # Copie les headers par défaut
    "X-Bin-Private": "false",  # ⚠️ IMPORTANT : rend le bin PUBLIC
    "X-Bin-Name": "MonProfil"  # Nom optionnel pour retrouver le bin
}

print("📤 Envoi de la requête POST...")
print(f"   URL : {BASE_URL}")
print(f"   Données : {json.dumps(mon_profil, indent=2, ensure_ascii=False)}")
print()

# Envoi de la requête POST
response = requests.post(
    BASE_URL,
    json=mon_profil,
    headers=headers_creation
)

# Analyse de la réponse
print(f"📊 Code HTTP : {response.status_code}")

if response.status_code == 200:
    result = response.json()
    bin_id = result["metadata"]["id"]
    
    print("\n✅ BIN CRÉÉ AVEC SUCCÈS !")
    print("="*60)
    print(f"🆔 ID du Bin : {bin_id}")
    print(f"\n🌐 URL pour voir dans le NAVIGATEUR :")
    print(f"   https://api.jsonbin.io/v3/b/{bin_id}")
    print("\n💡 Copie cette URL et ouvre-la dans ton navigateur !")
    print("="*60)
    
    # Sauvegarder l'ID pour les prochaines cellules
    BIN_ID = bin_id
else:
    print(f"\n❌ ERREUR : {response.text}")
    BIN_ID = None
```

### 🎯 Action : Ouvre l'URL dans ton navigateur !

Tu devrais voir tes données JSON affichées. C'est la preuve que :
1. Ta requête POST a fonctionné
2. Les données sont réellement stockées sur le serveur
3. Le bin est accessible publiquement

## ✏️ Étape 4 : Modifier le Bin avec PUT

Modifions les données existantes. PUT remplace TOUTES les données du bin.

```python
# ============================================
# ✏️ MODIFIER UN BIN (PUT)
# ============================================

if BIN_ID is None:
    print("❌ Aucun Bin ID disponible. Exécute d'abord la cellule POST !")
else:
    # Nouvelles données (modification)
    profil_mis_a_jour = {
        "nom": "Ton Nom",
        "prenom": "Ton Prénom",
        "formation": "Python & APIs",
        "niveau": "Intermédiaire",  # ← Niveau modifié !
        "competences": ["Python", "HTTP", "REST", "JSON", "Authentification API"],  # ← Nouvelle compétence !
        "date_creation": datetime.now().isoformat(),
        "date_modification": datetime.now().isoformat(),  # ← Nouveau champ !
        "message": "Bin mis à jour via PUT ! 🚀",
        "exercices_completes": 4  # ← Nouveau champ !
    }
    
    url_modification = f"{BASE_URL}/{BIN_ID}"
    
    print(f"✏️ Modification du Bin : {BIN_ID}")
    print(f"   URL : {url_modification}")
    print()
    
    # Requête PUT
    response = requests.put(
        url_modification,
        json=profil_mis_a_jour,
        headers=HEADERS
    )
    
    print(f"📊 Code HTTP : {response.status_code}")
    
    if response.status_code == 200:
        result = response.json()
        
        print("\n✅ BIN MIS À JOUR AVEC SUCCÈS !")
        print("="*60)
        print(json.dumps(result["record"], indent=2, ensure_ascii=False))
        print("="*60)
        print(f"\n🌐 Rafraîchis l'URL dans ton navigateur pour voir les changements :")
        print(f"   https://api.jsonbin.io/v3/b/{BIN_ID}")
    else:
        print(f"\n❌ ERREUR : {response.text}")
```

```python
# ✅ CORRIGÉ EXERCICE 3
# ============================================

class JSONBinClient:
    """Client pour interagir avec l'API JSONBin.io"""
    
    BASE_URL = "https://api.jsonbin.io/v3/b"
    
    def __init__(self, api_key):
        self.api_key = api_key
        self.headers = {
            "Content-Type": "application/json",
            "X-Master-Key": api_key
        }
    
    def create(self, data, name=None, public=True):
        """Crée un nouveau bin et retourne son ID"""
        headers = {
            **self.headers,
            "X-Bin-Private": "false" if public else "true"
        }
        if name:
            headers["X-Bin-Name"] = name
        
        try:
            response = requests.post(self.BASE_URL, json=data, headers=headers)
            response.raise_for_status()
            result = response.json()
            bin_id = result["metadata"]["id"]
            print(f"✅ Bin créé : {bin_id}")
            print(f"🌐 URL : https://api.jsonbin.io/v3/b/{bin_id}")
            return bin_id
        except requests.exceptions.HTTPError as e:
            print(f"❌ Erreur création : {e}")
            return None
    
    def read(self, bin_id):
        """Lit et retourne les données d'un bin"""
        try:
            response = requests.get(f"{self.BASE_URL}/{bin_id}", headers=self.headers)
            response.raise_for_status()
            return response.json()["record"]
        except requests.exceptions.HTTPError as e:
            print(f"❌ Erreur lecture : {e}")
            return None
    
    def update(self, bin_id, data):
        """Met à jour un bin existant"""
        try:
            response = requests.put(f"{self.BASE_URL}/{bin_id}", json=data, headers=self.headers)
            response.raise_for_status()
            print(f"✅ Bin {bin_id} mis à jour !")
            return response.json()["record"]
        except requests.exceptions.HTTPError as e:
            print(f"❌ Erreur mise à jour : {e}")
            return None
    
    def delete(self, bin_id):
        """Supprime un bin"""
        try:
            response = requests.delete(f"{self.BASE_URL}/{bin_id}", headers=self.headers)
            response.raise_for_status()
            print(f"✅ Bin {bin_id} supprimé !")
            return True
        except requests.exceptions.HTTPError as e:
            print(f"❌ Erreur suppression : {e}")
            return False

print("✅ Classe JSONBinClient créée !")
```

## 🎯 Exercice Final : Mini-Application de Notes

Crée une application complète de gestion de notes avec :
1. `ajouter_note(titre, contenu)` → Ajoute une note
2. `lister_notes()` → Affiche toutes les notes
3. `modifier_note(id, nouveau_contenu)` → Modifie une note
4. `supprimer_note(id)` → Supprime une note

Toutes les notes doivent être stockées dans un seul bin JSONBin !

```python
# ✅ EXERCICE FINAL : Application de Notes
# ============================================

class NotesApp:
    """Application de gestion de notes avec JSONBin.io"""
    
    def __init__(self, api_key):
        self.client = JSONBinClient(api_key)
        self.bin_id = None
        self.notes = []
        self.next_id = 1
    
    def initialiser(self, bin_id=None):
        """Initialise l'application (crée ou charge un bin existant)"""
        if bin_id:
            # Charger un bin existant
            self.bin_id = bin_id
            data = self.client.read(bin_id)
            if data:
                self.notes = data.get("notes", [])
                self.next_id = data.get("next_id", 1)
                print(f"✅ {len(self.notes)} notes chargées !")
        else:
            # Créer un nouveau bin
            data = {"notes": [], "next_id": 1, "created_at": datetime.now().isoformat()}
            self.bin_id = self.client.create(data, name="MesNotes")
            print("✅ Nouvelle application de notes créée !")
        
        return self.bin_id
    
    def _sauvegarder(self):
        """Sauvegarde les notes dans JSONBin"""
        data = {
            "notes": self.notes,
            "next_id": self.next_id,
            "updated_at": datetime.now().isoformat()
        }
        self.client.update(self.bin_id, data)
    
    def ajouter_note(self, titre, contenu):
        """Ajoute une nouvelle note"""
        note = {
            "id": self.next_id,
            "titre": titre,
            "contenu": contenu,
            "created_at": datetime.now().isoformat()
        }
        self.notes.append(note)
        self.next_id += 1
        self._sauvegarder()
        print(f"✅ Note #{note['id']} ajoutée : {titre}")
        return note["id"]
    
    def lister_notes(self):
        """Affiche toutes les notes"""
        if not self.notes:
            print("📋 Aucune note pour le moment.")
            return
        
        print("\n📋 MES NOTES :")
        print("="*50)
        for note in self.notes:
            print(f"\n📝 Note #{note['id']} : {note['titre']}")
            print(f"   {note['contenu']}")
            print(f"   📅 {note['created_at'][:10]}")
        print("="*50)
    
    def modifier_note(self, note_id, nouveau_titre=None, nouveau_contenu=None):
        """Modifie une note existante"""
        for note in self.notes:
            if note["id"] == note_id:
                if nouveau_titre:
                    note["titre"] = nouveau_titre
                if nouveau_contenu:
                    note["contenu"] = nouveau_contenu
                note["updated_at"] = datetime.now().isoformat()
                self._sauvegarder()
                print(f"✅ Note #{note_id} modifiée !")
                return True
        print(f"❌ Note #{note_id} non trouvée.")
        return False
    
    def supprimer_note(self, note_id):
        """Supprime une note"""
        for i, note in enumerate(self.notes):
            if note["id"] == note_id:
                self.notes.pop(i)
                self._sauvegarder()
                print(f"✅ Note #{note_id} supprimée !")
                return True
        print(f"❌ Note #{note_id} non trouvée.")
        return False
    
    def get_url(self):
        """Retourne l'URL pour voir les notes dans le navigateur"""
        return f"https://api.jsonbin.io/v3/b/{self.bin_id}"

print("✅ Classe NotesApp créée !")
```
