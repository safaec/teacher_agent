 Lecture de fichiers JSON

JSON (JavaScript Object Notation) est le format standard des APIs et du web.

````python
# JSON plat (tableau d'objets)
df = pd.read_json("OpenWeatherMap_simple.json")
print("Fichier OpenWeatherMap_simple")
display(df)
````

````python
# Exemple avec JSON imbriqué

# JSON imbriqué - nécessite normalisation
import json

# D'abord, chargez le JSON comme un dictionnaire
with open('OpenWeatherMap_imbrique.json', 'r') as f:
    data = json.load(f)
# r est le mode de lecture du fichier 
# f est le nom du fichier ouvert 

# Utilisez json_normalize pour aplatir la structure imbriquée
df_complet = pd.json_normalize(data, sep='_')
# sep définit le séparateur pour les noms de colonnes imbriquées
# _ est souvent utilisé pour une meilleure lisibilité des noms de colonnes 

print("DataFrame complètement aplati:")
print(f"Shape: {df_complet.shape}")
print(f"Colonnes: {df_complet.columns.tolist()}")
display(df_complet)
````

````python
# Extraire seulement les données météo
donnees_meteo = []
for item in data:
    donnees_meteo.append(item['weather'])

# Afficher les données météo brutes
print("\nDonnées météo brutes:")
display(donnees_meteo)

# Aplatir les données météo
df_meteo = pd.DataFrame(donnees_meteo)
print("\nDataFrame des données météo:")
display(df_meteo)

# Ajouter une colonne pour la ville correspondante
df_meteo['ville'] = [item['location']['city'] for item in data]


print("\nDonnées météo uniquement:")
display(df_meteo[['ville', 'main', 'temperature', 'feels_like', 'humidity']])
````

**Ligne 10 :** Convertit la liste de dictionnaires `donnees_meteo` en un DataFrame pandas nommé `df_meteo`. Cette transformation "aplatit" les données en les organisant dans un format tabulaire.

**Ligne 18 :** Affiche un titre "Données météo uniquement:" pour la section finale des résultats.

**Ligne 19-20 :** Sélectionne et affiche uniquement les 5 colonnes spécifiées du DataFrame, dans l'ordre demandé :

- `temperature` : la température actuelle mesurée
- `feels_like` : la température ressentie, qui peut différer de la température réelle

````python
# Extraire localisation + température

# Extraire ville, pays, température
localisation_temp = []
for item in data:
    localisation_temp.append({
        'ville': item['location']['city'],
        'pays': item['location']['country'],
        'temperature': item['weather']['temperature'],
        'latitude': item['location']['coordinates']['lat'],
        'longitude': item['location']['coordinates']['lon']
    })

df_local_temp = pd.DataFrame(localisation_temp)
print("\nLocalisation et température:")
display(df_local_temp)
````

> 💭 **Question Socratique 2** : Les fichiers Excel peuvent contenir des formules, des graphiques, et des mises en forme. Que pensez-vous qu'il arrive à ces éléments quand vous lisez le fichier avec pandas ? Sont-ils préservés ?

> 💭 **Réponse** 

👉 **Non, les formules, graphiques et mises en forme d’un fichier Excel ne sont pas préservés par pandas.**
Pandas ne lit **que les valeurs des cellules**, pas la logique ni la présentation.

Quand tu utilises `pandas.read_excel()` :

1. **Formules**

* ❌ La **formule elle-même** (`=SUM(A1:A10)`) est **perdue**
* ✅ Pandas récupère **le résultat calculé** de la formule (la valeur affichée dans Excel)
* 👉 Si le fichier n’a jamais été recalculé dans Excel, pandas peut lire une valeur **obsolète**

📌 Interprétation data :
> pandas travaille sur un **snapshot des données**, pas sur la logique de calcul.


 2. **Graphiques**

* ❌ **Totalement ignorés**
* Aucun objet graphique n’est chargé, analysable ou modifiable
* Pandas ne “voit” que les feuilles et les cellules

📌 Normal :
> Un graphique est une **visualisation**, pas une donnée brute.


 3. **Mises en forme (couleurs, gras, bordures, formats conditionnels)**

* ❌ **Perdues**
* Pandas lit le **contenu**, pas la présentation
* Une cellule rouge = juste une valeur pour pandas

📌 Conséquence importante :
> Toute information portée uniquement par la couleur ou le style est **perdue** → très mauvais design data.


 Pourquoi c’est logique ?
 
Pandas est une bibliothèque de :
* manipulation de données
* calcul
* analyse statistique

👉 **Pas un moteur Excel**, ni un outil de reporting.

Il convertit Excel → **DataFrame**
Et un DataFrame = valeurs + types, rien d’autre.

💡 **Excel ≠ source “data-safe” quand la logique est visuelle ou implicite**

* Une donnée doit être **explicite**, pas encodée par :
  * une couleur
  * une formule cachée
  * un graphique


> *Quand on lit un fichier Excel avec pandas, seules les valeurs finales des cellules sont conservées. Les formules, graphiques et mises en forme sont ignorés, car pandas ne manipule que des données, pas leur logique ni leur présentation.*



 Lecture de fichiers volumineux (Chunking)

Quand un fichier dépasse la mémoire disponible, le **chunking** est la solution.

 Le problème

```python
 ❌ Ceci peut faire planter votre machine avec un fichier de 10 Go
df = pd.read_csv("enorme_fichier.csv")
```

 La solution : chunksize

````python
# ✅ Lecture par morceaux de 100 000 lignes
chunks = pd.read_csv("enorme_fichier.csv", chunksize=100_000)

# Traitement itératif
resultats = []
for chunk in chunks:
   #Traitement sur chaque chunk
    resultat = chunk.groupby("categorie")["ventes"].sum()
    resultats.append(resultat)

# Combinaison des résultats
final = pd.concat(resultats).groupby(level=0).sum()
````

 Quand utiliser le chunking ?

| Taille du fichier | RAM disponible | Recommandation |
|-|-|-|
| < 500 Mo | 8 Go | Lecture directe |
| 500 Mo - 2 Go | 8 Go | `usecols` + types optimisés |
| > 2 Go | 8 Go | **Chunking obligatoire** |
| > 10 Go | Toute RAM | Considérer Dask ou Spark |

*(Source : [pandas Documentation - Scaling](https://pandas.pydata.org/docs/user_guide/scale.html))*
